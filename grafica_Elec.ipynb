## Actividad electiva - grafica.

```python
x = np.linspace(0,80,700)
y = np.sin(x)

plt.plot(x,y, color = "darkviolet")

plt.title("F(x) = sen(x)")
plt.xlabel("x")
plt.ylabel("sen(x)")

plt.xlim(0,40)
plt.ylim(-2,2)

plt.grid()
plt.show()
```
